```python
import pandas as pd
import io
import requests

# URLs for the series
export_url = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=EXP5330"
import_url = "https://fred.stlouisfed.org/graph/fredgraph.csv?id=IMP5330"

# Download
export_df = pd.read_csv(io.StringIO(requests.get(export_url).text))
import_df = pd.read_csv(io.StringIO(requests.get(import_url).text))

# Rename columns
export_df.rename(columns={export_df.columns[0]: "Date", export_df.columns[1]: "US_Exports_to_India"}, inplace=True)
import_df.rename(columns={import_df.columns[0]: "Date", import_df.columns[1]: "US_Imports_from_India"}, inplace=True)

# Convert dates
export_df["Date"] = pd.to_datetime(export_df["Date"], errors="coerce")
import_df["Date"] = pd.to_datetime(import_df["Date"], errors="coerce")

# Merge
trade_df = pd.merge(export_df, import_df, on="Date", how="outer").sort_values("Date").reset_index(drop=True)

# Filter from Oct 2000 to Oct 2025
start = pd.to_datetime("2000-10-01")
end   = pd.to_datetime("2025-10-31")
trade_df = trade_df[(trade_df["Date"] >= start) & (trade_df["Date"] <= end)]

# Interpolate missing values
trade_df.interpolate(method="linear", inplace=True)

# Save
trade_df.to_csv("US_India_TradeFlow_2000Oct_to_2025Oct.csv", index=False)
print("✅ Saved trade dataset from Oct 2000 to Oct 2025.")
print(trade_df.head(), trade_df.tail())

```

Merge All Market Datasets

```python
import pandas as pd

# File list
files = {
    "djia": "djia_2000_2025.csv",
    "nifty": "nifty_50_2000_2025.csv",
    "india_vix": "india_vix_2000_2025.csv",
    "us_vix": "us_vix_2000_2025.csv",
    "usd_inr": "usd_inr_2000_2025.csv"
}

dataframes = {}

for name, path in files.items():
    df = pd.read_csv(path)
    df.columns = df.columns.str.strip().str.lower()

    # --- Step 1: Detect date column automatically ---
    date_col = None
    for c in df.columns:
        if 'date' in c or 'time' in c or 'day' in c:
            date_col = c
            break

    # --- Special fix for india_vix (its 'price' column is the date) ---
    if name == "india_vix" and date_col is None and 'price' in df.columns:
        date_col = 'price'

    # --- Step 2: If still not found, try first column ---
    if date_col is None:
        date_col = df.columns[0]

    # --- Step 3: Rename that column to 'date' ---
    df.rename(columns={date_col: 'date'}, inplace=True)

    # --- Step 4: Convert to datetime safely ---
    df['date'] = pd.to_datetime(df['date'], errors='coerce')
    df.dropna(subset=['date'], inplace=True)
    df.sort_values('date', inplace=True)

    dataframes[name] = df
    print(f"✅ {name} → Date column used: '{date_col}' | Columns: {df.columns.tolist()}")

# --- Step 5: Merge all on 'date' ---
merged = (
    dataframes["djia"]
    .merge(dataframes["nifty"], on="date", how="inner", suffixes=("_djia", "_nifty"))
    .merge(dataframes["india_vix"], on="date", how="inner")
    .merge(dataframes["us_vix"], on="date", how="inner", suffixes=("_india_vix", "_us_vix"))
    .merge(dataframes["usd_inr"], on="date", how="inner")
)

merged = merged.loc[:, ~merged.columns.duplicated()]
merged.to_csv("merged_stock_data_2000_2025.csv", index=False)

print("\n🎯 Final merged dataset shape:", merged.shape)
print("💾 Saved as: merged_stock_data_2000_2025.csv")
print(merged.head())
```

Preprocessing of merged file

```python
import pandas as pd
import numpy as np

# --- Load dataset ---
merged = pd.read_csv("merged_stock_data_2000_2025.csv")

print("🔹 Before Cleaning:")
print(f"Rows: {merged.shape[0]}, Columns: {merged.shape[1]}")

# --- Step 1: Ensure date column is datetime ---
if 'date' in merged.columns:
    merged['date'] = pd.to_datetime(merged['date'], errors='coerce')
    merged.dropna(subset=['date'], inplace=True)
else:
    print("⚠️ No 'date' column found!")

# --- Step 2: Fill missing numeric values with median (not 0) ---
for col in merged.select_dtypes(include=[np.number]).columns:
    merged[col].fillna(merged[col].median(), inplace=True)

# --- Step 3: Handle outliers using IQR ---
numeric_cols = merged.select_dtypes(include=[np.number]).columns
for col in numeric_cols:
    Q1 = merged[col].quantile(0.25)
    Q3 = merged[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = merged[(merged[col] < lower) | (merged[col] > upper)].shape[0]

    if outliers > 0:
        median_val = merged[col].median()
        merged.loc[merged[col] < lower, col] = median_val
        merged.loc[merged[col] > upper, col] = median_val
        print(f"✅ {col}: {outliers} outliers replaced with median ({median_val})")

print("\n✅ After Cleaning:")
print(f"Rows: {merged.shape[0]}, Columns: {merged.shape[1]}")
print("Missing values per column:\n", merged.isna().sum())

# --- Save clean version ---
merged.to_csv("cleaned_merged_stock_data.csv", index=False)
print("\n💾 Cleaned merged dataset saved as: cleaned_merged_stock_data.csv")
```

```python
# Load Trade Flow dataset
trade = pd.read_csv("/content/US_India_TradeFlow_2000Oct_to_2025Oct.csv")
from scipy import stats
print("\nBefore Cleaning (Trade Flow):")
print(trade.info())
print("\nMissing values per column:")
print(trade.isnull().sum())

# --- Step 1: Standardize column names ---
trade.columns = trade.columns.str.strip().str.lower()

# --- Step 2: Detect date column ---
date_col = None
for c in trade.columns:
    if 'date' in c or 'year' in c or 'month' in c or 'time' in c:
        date_col = c
        break
if date_col is None:
    date_col = trade.columns[0]  # fallback

trade.rename(columns={date_col: 'date'}, inplace=True)
trade['date'] = pd.to_datetime(trade['date'], errors='coerce')
trade.dropna(subset=['date'], inplace=True)

# --- Step 3: Fill missing numeric data ---
trade = trade.fillna(method='ffill').fillna(method='bfill')

# --- Step 4: Handle Outliers (z-score method) ---
numeric_cols = trade.select_dtypes(include=[np.number]).columns
z_scores = np.abs(stats.zscore(trade[numeric_cols], nan_policy='omit'))
trade = trade[(z_scores < 3).all(axis=1)]

# --- Step 5: Save cleaned data ---
trade.to_csv("cleaned_trade_flow_data.csv", index=False)

print("\n✅ Cleaned Trade Flow Data saved as 'cleaned_trade_flow_data.csv'")
print(f"Rows: {trade.shape[0]}, Columns: {trade.shape[1]}")
print("Columns:", list(trade.columns))
```

Perform Granger Causality Testing

```python
from statsmodels.tsa.stattools import grangercausalitytests
import pandas as pd

# Pick columns the same way as your code
us_col = close_cols[0] + "_log_return"
inr_col = close_cols[1] + "_log_return"

# Prepare dataset
gc_data = closing_only[[inr_col, us_col]].dropna()

# Function to extract only p-values
def clean_granger(data, maxlag):
    results = {}
    for lag in range(1, maxlag + 1):
        test = grangercausalitytests(data, lag, verbose=False)
        p_val = test[lag][0]['ssr_chi2test'][1]   # extract p-value
        results[f"Lag {lag}"] = round(p_val, 5)
    return pd.DataFrame.from_dict(results, orient='index', columns=['p-value'])

# ---------------------------
# 1. US → INR
# ---------------------------
print("---------------------------------------------------")
print("   US → INR Granger Causality (Clean Output)")
print("---------------------------------------------------")
clean_us_in = clean_granger(gc_data, 5)
print(clean_us_in)

# ---------------------------
# 2. INR → US
# ---------------------------
print("\n---------------------------------------------------")
print("   INR → US Granger Causality (Clean Output)")
print("---------------------------------------------------")
gc_data_rev = closing_only[[us_col, inr_col]].dropna()
clean_in_us = clean_granger(gc_data_rev, 5)
print(clean_in_us)
```

```python
# ===============================================================
# 0. IMPORTS
# ===============================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# ===============================================================
# 1. USE YOUR CORRECT TARGET COLUMN
# ===============================================================
target_col = "close_nifty"

# df = your dataframe (already loaded)
df = df[[target_col]].copy()
df = df.dropna()

# ===============================================================
# 2. TRAIN/TEST SPLIT + SCALING
# ===============================================================
sequence_length = 60
test_size = 0.2

n_rows = len(df)
train_rows = int((1 - test_size) * n_rows)

# Fit scaler ONLY on training data (very important)
scaler = MinMaxScaler()
scaler.fit(df.iloc[:train_rows])

scaled_all = scaler.transform(df)

# ===============================================================
# 3. CREATE SEQUENCES
# ===============================================================
def create_sequences(data, seq_len):
    X, y = [], []
    for i in range(seq_len, len(data)):
        X.append(data[i-seq_len:i, 0])
        y.append(data[i, 0])
    X = np.array(X).reshape(-1, seq_len, 1)
    y = np.array(y)
    return X, y

X, y = create_sequences(scaled_all, sequence_length)

first_seq_index = sequence_length
train_count = train_rows - first_seq_index

X_train = X[:train_count]
y_train = y[:train_count]
X_test = X[train_count:]
y_test = y[train_count:]

# Align dates with sequences
dates = df.index[first_seq_index:]
test_dates = dates[train_count:]

# ===============================================================
# 4. LSTM MODEL
# ===============================================================
model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(sequence_length, 1)),
    Dropout(0.2),

    LSTM(32),
    Dropout(0.2),

    Dense(1)
])

model.compile(optimizer="adam", loss="mse")

es = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)

history = model.fit(
    X_train, y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.1,
    callbacks=[es],
    verbose=1
)

# ===============================================================
# 5. PREDICTION + INVERSE TRANSFORM
# ===============================================================
y_pred_scaled = model.predict(X_test)

y_pred = scaler.inverse_transform(y_pred_scaled)
y_test_orig = scaler.inverse_transform(y_test.reshape(-1,1))

# ===============================================================
# 6. IMPROVED MAIN PLOT (VISIBLE DIFFERENCE)
# ===============================================================
plt.figure(figsize=(14,7))

plt.plot(
    dates,
    scaler.inverse_transform(y.reshape(-1,1)),
    label='Actual NIFTY (Train+Test)',
    color='blue',
    linewidth=1.8,
    alpha=0.5
)

plt.plot(
    test_dates,
    y_test_orig,
    label='Actual NIFTY (Test)',
    color='navy',
    linewidth=2.5
)

plt.plot(
    test_dates,
    y_pred,
    label='Predicted NIFTY (LSTM)',
    color='red',
    linestyle='--',
    linewidth=2.8
)

plt.title("LSTM Prediction vs Actual — NIFTY 50 (close_nifty)", fontsize=16)
plt.xlabel("Date", fontsize=13)
plt.ylabel("Price", fontsize=13)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()

# ===============================================================
# 7. ZOOMED PLOT FOR CRYSTAL-CLEAR SEPARATION
# ===============================================================
plt.figure(figsize=(14,7))

plt.plot(test_dates, y_test_orig, label='Actual Test Data', color='blue', linewidth=2.5)
plt.plot(test_dates, y_pred, label='LSTM Prediction', color='red', linestyle='--', linewidth=2.8)

plt.title("Zoomed — LSTM vs Actual (Test Region)", fontsize=16)
plt.xlabel("Date", fontsize=13)
plt.ylabel("Price", fontsize=13)
plt.grid(True, alpha=0.3)
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()
```

Full ARIMA Workflow for Forecasting All Closing Price Series (2000–2025)

```python
# ===========================================
#       ARIMA MODEL FOR ALL CLOSING PRICES
# ===========================================

import pandas as pd
import numpy as np
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error, mean_absolute_error
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

# ----------------------------
# Load Dataset
# ----------------------------
df = pd.read_csv("/content/closing_price_dataset_2000_2025.csv")  # <-- Replace with your file path
df['date'] = pd.to_datetime(df['date'])
df.set_index('date', inplace=True)

print("Columns in dataset:", df.columns)


# ----------------------------
# ARIMA FUNCTION
# ----------------------------
def run_arima(series, col_name):

    print(f"\n======================================")
    print(f"     ARIMA MODEL RESULTS → {col_name} ")
    print(f"======================================")

    # Train-test split (80%-20%)
    train_size = int(len(series) * 0.80)
    train, test = series[:train_size], series[train_size:]

    # Fit ARIMA(5,1,2)
    model = ARIMA(train, order=(5,1,2))
    model_fit = model.fit()

    # Forecast for same length as test
    forecast = model_fit.forecast(steps=len(test))

    # Performance Metrics
    rmse = np.sqrt(mean_squared_error(test, forecast))
    mae = mean_absolute_error(test, forecast)

    print(f"RMSE: {rmse}")
    print(f"MAE: {mae}")

    # ----------------------------
    # Plot 1 — Full Training + Test
    # ----------------------------
    plt.figure(figsize=(12,5))
    plt.plot(train.index, train, label="Training Data")
    plt.plot(test.index, test, label="Actual Test Data")
    plt.plot(test.index, forecast, label="ARIMA Forecast")
    plt.title(f"ARIMA Forecast — {col_name}")
    plt.legend()
    plt.grid(True)
    plt.show()

    # ----------------------------
    # Plot 2 — Zoomed Test vs Forecast
    # ----------------------------
    plt.figure(figsize=(12,5))
    plt.plot(test.index, test, label="Actual Test Data")
    plt.plot(test.index, forecast, label="Forecast")
    plt.title(f"Zoom View — Test Period Only ({col_name})")
    plt.legend()
    plt.grid(True)
    plt.show()

    return rmse, mae


# ---------------------------------------------
# Run ARIMA on all closing price columns
# ---------------------------------------------
cols = [
    'close_djia',
    'close_nifty',
    'close_india_vix',
    'close_us_vix',
    'close_usd_inr'
]

results = {}

for col in cols:
    series = df[col].dropna()
    rmse, mae = run_arima(series, col)
    results[col] = {"RMSE": rmse, "MAE": mae}

# ---------------------------------------------
# Final Summary Table
# ---------------------------------------------
summary_df = pd.DataFrame(results).T
print("\n======================================")
print("         ARIMA PERFORMANCE TABLE")
print("======================================")
print(summary_df)
```

```python
# -------------------------
# Analysis + GARCH pipeline
# -------------------------
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from statsmodels.tsa.stattools import adfuller, acf, pacf
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.stats.diagnostic import acorr_ljungbox, het_arch
from arch import arch_model

# ---------- SETTINGS ----------
target_col = "close_nifty"   # your target column
# --------------------------------

# ---- 0. Basic checks ----
assert target_col in df.columns, f"Column {target_col} not found in df.columns: {df.columns.tolist()}"
# ensure datetime index
if not isinstance(df.index, pd.DatetimeIndex):
    # try to coerce if a date column exists
    if 'date' in df.columns:
        df['date'] = pd.to_datetime(df['date'])
        df = df.set_index('date')
    else:
        raise ValueError("DataFrame must have a DatetimeIndex or a 'date' column")

# keep only target and drop NA
series = df[[target_col]].dropna().copy()
series.columns = ['price']

print("Data loaded. Range:", series.index.min(), "to", series.index.max())
print("Length:", len(series))

# ---- 1. Plot price ----
plt.figure(figsize=(12,5))
plt.plot(series.index, series['price'], label='Close (NIFTY)')
plt.title('NIFTY Close (close_nifty)')
plt.xlabel('Date')
plt.ylabel('Price')
plt.grid(alpha=0.3)
plt.legend()
plt.show()

# ---- 2. Compute log returns ----
series['log_ret'] = np.log(series['price']).diff().dropna()
rets = series['log_ret'].dropna()

plt.figure(figsize=(12,4))
plt.plot(rets.index, rets, label='Log Returns')
plt.title('Log Returns')
plt.grid(alpha=0.3)
plt.legend()
plt.show()

# Summary stats
print("Returns summary:\n", rets.describe().round(6))

# ---- 3. Stationarity test (ADF) ----
adf_res = adfuller(rets.dropna())
print("\nADF test on log returns:")
print(f"  ADF stat = {adf_res[0]:.4f}, p-value = {adf_res[1]:.4f}")
for k, v in zip(['Test Stat','p-value','Lags used','Number of obs'], adf_res[:4]):
    print(" ", k, ":", v)

# ---- 4. Check for ARCH effects ----
# Engle's ARCH test
arch_test = het_arch(rets.dropna(), nlags=12)  # try 12 lags
print("\nEngle's ARCH test (12 lags):")
print("  LM stat:", arch_test[0], "p-value:", arch_test[1])

# Ljung-Box on squared returns (test for remaining autocorr in volatility)
lb = acorr_ljungbox(rets.dropna()**2, lags=[10, 20], return_df=True)
print("\nLjung-Box on squared returns:")
print(lb)

# ---- 5. ACF/PACF for returns and squared returns ----
plt.figure(figsize=(12,4))
plot_acf(rets.dropna(), lags=40, title='ACF of returns')
plt.show()

plt.figure(figsize=(12,4))
plot_pacf(rets.dropna(), lags=40, title='PACF of returns')
plt.show()

plt.figure(figsize=(12,4))
plot_acf((rets.dropna())**2, lags=40, title='ACF of squared returns (for ARCH)')
plt.show()

# ---- 6. Fit GARCH(1,1) to returns ----
# We'll model mean as zero or small AR(1) if desired. Common approach: mean='Zero' or 'Constant' or include AR terms.
# Here: mean='Constant' and volatility model GARCH(1,1) with Student-t errors (more robust to fat tails)
am = arch_model(rets*100,  # scale returns to percent to help numeric stability
                mean='Constant', vol='GARCH', p=1, q=1, dist='StudentsT')

res = am.fit(update_freq=10, disp='off')
print("\nGARCH(1,1) fit summary:")
print(res.summary())

# ---- 7. Diagnostics plots ----
# conditional volatility (annualization depends on freq; we will just plot the estimated sigma)
cond_vol = res.conditional_volatility / 100.0  # back to log-return scale
plt.figure(figsize=(12,5))
plt.plot(cond_vol.index, cond_vol, label='Estimated cond vol (sigma)')
plt.title('Estimated Conditional Volatility (GARCH(1,1))')
plt.xlabel('Date')
plt.ylabel('Sigma (log-returns)')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# Standardized residuals and their ACF
std_resid = res.std_resid.dropna()
plt.figure(figsize=(12,4))
plt.plot(std_resid.index, std_resid, label='Std Residuals')
plt.title('Standardized Residuals')
plt.grid(alpha=0.3)
plt.show()

plt.figure(figsize=(12,4))
plot_acf(std_resid, lags=40, title='ACF of Standardized Residuals')
plt.show()

plt.figure(figsize=(12,4))
plot_acf(std_resid**2, lags=40, title='ACF of squared standardized residuals')
plt.show()

# ---- 8. Forecasting volatility ----
# 1-step ahead and horizon
horizon = 10
fore = res.forecast(horizon=horizon, reindex=False)
# The forecast object contains variance forecasts. We'll take the last available forecast for t+1..t+h
last_mean_fore = fore.mean.iloc[-1]
last_var_fore = fore.variance.iloc[-1]
print(f"\nNext {horizon} steps forecast (mean returns *100):")
print(last_mean_fore)
print("\nNext {0} steps forecast variance (*100^2):".format(horizon))
print(last_var_fore)

# Convert variance -> sigma (in original log-return units)
sigma_fore = np.sqrt(last_var_fore) / 100.0
print("\nForecast sigma (log-return scale):")
print(sigma_fore)

# Plot variance forecast
plt.figure(figsize=(10,4))
plt.plot(np.arange(1, horizon+1), sigma_fore.values, marker='o')
plt.title('Forecast volatility (sigma) for next %d steps' % horizon)
plt.xlabel('Steps ahead')
plt.ylabel('Sigma (log-returns)')
plt.grid(alpha=0.3)
plt.show()

# ---- 9. Save outputs ----
# ---- 9. Save outputs ----
import pickle

with open("garch_fit_results.pkl", "wb") as f:
    pickle.dump(res, f)

print("\nModel saved to garch_fit_results.pkl")

# End of pipeline
```

```python
# ---------- CLEANED IRF + FEVD ----------

# choose horizons
irf_horizon = 10    # IRF horizon (0..10)
fevd_horizon = 5    # FEVD horizon (0..5)

# IRF (safe)
irf = res.irf(irf_horizon)

# ensure names are available
names = list(res.names)
print("VAR variable names:", names)

# summary function for IRF impulse -> response
def irf_inspect(impulse, response, thresh_ratio=0.10):
    if impulse not in names or response not in names:
        print(f"Skipping IRF {impulse}->{response} (variable missing).")
        return None
    imp_idx = names.index(impulse)
    resp_idx = names.index(response)
    series = irf.irfs[:, resp_idx, imp_idx]   # length = irf_horizon + 1
    print(f"\nIRF {impulse} -> {response} (t=0..{irf_horizon}):")
    print(np.round(series, 6).tolist())
    # peak info
    peak_day = int(np.argmax(np.abs(series)))
    peak_val = series[peak_day]
    sign = 'positive' if peak_val > 0 else ('negative' if peak_val < 0 else 'zero')
    print(f" Peak day: {peak_day}, peak value: {peak_val:.6g}, sign: {sign}")
    # duration: last day where abs(response) > thresh_ratio * abs(peak)
    if abs(peak_val) == 0:
        duration = 0
    else:
        thresh = thresh_ratio * abs(peak_val)
        above = np.where(np.abs(series) > thresh)[0]
        duration = int(above[-1]) if len(above) > 0 else 0
    print(f" Duration (last day > {thresh_ratio*100:.0f}% of peak): day {duration}")
    return {'series': series, 'peak_day': peak_day, 'peak_value': peak_val, 'sign': sign, 'duration': duration}

# run IRF summaries for main channels if present
irf_results = {}
if 'DJIA' in names and 'NIFTY' in names:
    irf_results['DJIA->NIFTY'] = irf_inspect('DJIA','NIFTY')
if 'USDINR' in names and 'NIFTY' in names:
    irf_results['USDINR->NIFTY'] = irf_inspect('USDINR','NIFTY')
if 'USVIX' in names and 'NIFTY' in names:
    irf_results['USVIX->NIFTY'] = irf_inspect('USVIX','NIFTY')
if 'TRADE' in names and 'NIFTY' in names:
    irf_results['TRADE->NIFTY'] = irf_inspect('TRADE','NIFTY')

# optional: plot IRFs for available pairs
for key in irf_results.keys():
    imp, resp = key.split('->')
    irf.plot(impulse=imp, response=resp, orth=False)
    plt.title(f"IRF: {imp} → {resp}")
    plt.show()

# ---------- FEVD (safe) ----------
fevd = res.fevd(fevd_horizon)
print("\nFEVD decomposition array shape:", fevd.decomp.shape, "(horizons, neqs, neqs)")

# use res.names for variables
if 'NIFTY' in names:
    nifty_idx = names.index('NIFTY')
    available_horizons = fevd.decomp.shape[0]   # safe number of horizons available
    for hh in range(available_horizons):
        print(f"\nFEVD at horizon {hh}:")
        for var in names:
            var_idx = names.index(var)
            contrib_pct = fevd.decomp[hh, nifty_idx, var_idx] * 100
            print(f"  {var:<10} explains {contrib_pct:6.2f}% of NIFTY variance")
else:
    print("NIFTY not found in VAR.names — skipping FEVD for NIFTY.")
```

```python
# =======================================================
# FULL PIPELINE TO ESTIMATE:
#  - DIRECTION (Granger causality)
#  - LAG DURATION (IRF peak + decay)
#  - STRENGTH (FEVD % variance explained)
# =======================================================

import numpy as np
import pandas as pd
from statsmodels.tsa.api import VAR
from statsmodels.tsa.stattools import adfuller
import matplotlib.pyplot as plt

# -------------------------------------------------------
# 1) AUTO-DETECT REQUIRED COLUMNS
# -------------------------------------------------------
def find_col(df, keys):
    for k in keys:
        matches = [c for c in df.columns if k.lower() in c.lower()]
        if matches:
            return matches[0]
    return None

djia = find_col(df, ["djia", "dow", "dow jones"])
nifty = find_col(df, ["nifty"])
usd = find_col(df, ["usd", "usd/inr", "usd-inr", "usdinr"])
vix = find_col(df, ["vix", "us vix"])

print("Detected:")
print("DJIA:", djia)
print("NIFTY:", nifty)
print("USD/INR:", usd)
print("VIX:", vix)

# ensure all exist
needed = [djia, nifty, usd, vix]
if None in needed:
    raise ValueError("Missing one or more required columns.")

# -------------------------------------------------------
# 2) PREPARE RETURNS (stationary)
# -------------------------------------------------------
def to_returns(s):
    return np.log(s).diff().dropna()

df_r = pd.DataFrame({
    "DJIA": to_returns(df[djia]),
    "NIFTY": to_returns(df[nifty]),
    "USDINR": to_returns(df[usd]),
    "VIX": to_returns(df[vix])
}).dropna()

print("Prepared return dataset:", df_r.shape)

# -------------------------------------------------------
# 3) FIT VAR MODEL
# -------------------------------------------------------
model = VAR(df_r)
lag = model.select_order(12).selected_orders["aic"]
if lag is None:
    lag = 5

res = model.fit(lag)
print(res.summary())

# -------------------------------------------------------
# 4) DIRECTION = GRANGER CAUSALITY
# -------------------------------------------------------
print("\n==================== DIRECTION (GRANGER) ====================")

def gc(x, y):
    test = res.test_causality(x, [y], kind="f")
    print(f"{y} → {x}   | p-value = {test.pvalue:.5f}")

gc("NIFTY", "DJIA")
gc("DJIA", "NIFTY")
gc("NIFTY", "USDINR")
gc("NIFTY", "VIX")

# -------------------------------------------------------
# 5) LAG DURATION = IMPULSE RESPONSE
# -------------------------------------------------------
print("\n==================== LAG DURATION (IRF) ====================")

irf_h = 10
irf = res.irf(irf_h)

def irf_summary(imp, resp):
    imp_i = res.names.index(imp)
    resp_i = res.names.index(resp)
    series = irf.irfs[:, resp_i, imp_i]

    peak_day = int(np.argmax(np.abs(series)))
    peak_value = series[peak_day]
    sign = "positive" if peak_value > 0 else "negative"

    threshold = 0.1 * abs(peak_value)
    above = np.where(np.abs(series) > threshold)[0]
    duration = int(above[-1]) if len(above) else 0

    print(f"\nIRF {imp} → {resp}")
    print("Peak day:", peak_day)
    print("Peak value:", peak_value)
    print("Sign:", sign)
    print("Duration (decay):", duration)

    irf.plot(impulse=imp, response=resp)
    plt.show()

    return peak_day, duration

djia_lag = irf_summary("DJIA", "NIFTY")
usd_lag = irf_summary("USDINR", "NIFTY")
vix_lag = irf_summary("VIX", "NIFTY")

# -------------------------------------------------------
# 6) STRENGTH = FEVD (% variance explained)
# -------------------------------------------------------
print("\n==================== STRENGTH (FEVD) ====================")

fevd_h = 5
fevd = res.fevd(fevd_h)

n_idx = res.names.index("NIFTY")

for h in range(fevd.decomp.shape[0]):
    print(f"\nHorizon {h}:")
    for var in res.names:
        v_idx = res.names.index(var)
        contrib = fevd.decomp[h, n_idx, v_idx] * 100
        print(f"{var:7s} explains {contrib:.2f}% of NIFTY variance")

print("\n==================== DONE ====================")
```
